```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('cwfc.csv', sep=None, engine='python')

def get_column(i):
    s = df.iloc[:, i].astype(str).str.replace(',', '.').str.strip()
    return pd.to_numeric(s, errors='coerce').values

dates = df.iloc[:, 0].values
apple = get_column(1)
market = get_column(2)
vix = get_column(3)
gold = get_column(4)

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

data = pd.DataFrame({'Apple': apple, 'Market': market, 'VIX': vix, 'Gold': gold}).dropna()
returns = data.pct_change().dropna()

Y = returns[['Apple']]
X = returns[['Market', 'VIX', 'Gold']]

model = LinearRegression()
model.fit(X, Y)

print(f"Свободный член (Intercept): {model.intercept_[0]:.6f}")
print(f"Коэффициенты (Market, VIX, Gold): {model.coef_[0]}")

y_pred = model.predict(X)
print(f"R-squared: {r2_score(Y, y_pred):.4f}")
print(f"MAE: {mean_absolute_error(Y, y_pred):.6f}")
print(f"MSE: {mean_squared_error(Y, y_pred):.6f}")
```
