```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('cwfc.csv', sep=None, engine='python')

def get_column(i):
    s = df.iloc[:, i].astype(str).str.replace(',', '.').str.strip()
    return pd.to_numeric(s, errors='coerce').values

dates = df.iloc[:, 0].values
apple = get_column(1)
market = get_column(2)
vix = get_column(3)
gold = get_column(4)

data = pd.DataFrame({
    'Apple': apple,
    'Market': market,
    'VIX': vix,
    'Gold': gold
}).dropna()

returns = data.pct_change().dropna()

Y = returns['Apple'].values
X_factors = returns[['Market', 'VIX', 'Gold']].values

X = np.column_stack((np.ones(len(X_factors)), X_factors))

XTX = np.dot(X.T, X)
XTX_inv = np.linalg.inv(XTX)
XTY = np.dot(X.T, Y)
beta = np.dot(XTX_inv, XTY)

labels = ['Константа (w0)', 'Рынок (w1)', 'VIX (w2)', 'Золото (w3)']
print("РЕЗУЛЬТАТЫ РАСЧЕТА:")
for label, b in zip(labels, beta):
    print(f"{label}: {b:.6f}")
```
