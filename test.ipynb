```python
import pandas as pd
import numpy as np

df = pd.read_csv('cwfc.csv', sep=None, engine='python')

def get_clean_array(col_index):
    s = df.iloc[:, col_index].astype(str).str.replace(',', '.')
    return pd.to_numeric(s.str.extract(r'(\d+\.?\d*)')[0], errors='coerce').values

date_arr   = df.iloc[:, 0].values  # Даты (столбец 0)
apple_arr  = get_clean_array(1)    # Apple (столбец 1)
market_arr = get_clean_array(3)    # S&P 500 (столбец 3)
vix_arr    = get_clean_array(5)    # VIX (столбец 5)
gold_arr   = get_clean_array(7)    # Gold (столбец 7)

print("Данные успешно извлечены!")
print(f"Размер массива Apple: {len(apple_arr)} элементов")
print(f"Первые 5 значений Apple: {apple_arr[:5]}")
print(f"Первые 5 значений Рынка: {market_arr[:5]}")
print(f"Первые 5 значений VIX:    {vix_arr[:5]}")
print(f"Первые 5 значений Золота: {gold_arr[:5]}")
```
